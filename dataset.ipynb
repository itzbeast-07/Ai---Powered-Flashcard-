```python
import pandas as pd
import random

data = []

for _ in range(1000):

    rand = random.random()

    if rand < 0.3:
        score = random.randint(75, 100)
        time_taken = random.randint(30, 70)
        reattempts = random.randint(0, 1)
        topic_errors = random.randint(0, 2)
        avg_difficulty = round(random.uniform(0.7, 1.0), 2)
        consistency = round(random.uniform(0.7, 1.0), 2)
        level = "Advanced"

    elif rand < 0.7:
        score = random.randint(40, 74)
        time_taken = random.randint(60, 120)
        reattempts = random.randint(1, 3)
        topic_errors = random.randint(2, 5)
        avg_difficulty = round(random.uniform(0.5, 0.7), 2)
        consistency = round(random.uniform(0.5, 0.8), 2)
        level = "Intermediate"

    else:
        score = random.randint(0, 39)
        time_taken = random.randint(100, 180)
        reattempts = random.randint(2, 5)
        topic_errors = random.randint(5, 10)
        avg_difficulty = round(random.uniform(0.2, 0.5), 2)
        consistency = round(random.uniform(0.3, 0.6), 2)
        level = "Beginner"

    score = max(0, min(100, score + random.randint(-5, 5)))
    time_taken = max(20, time_taken + random.randint(-10, 10))
    consistency = round(min(1.0, max(0.2, consistency + random.uniform(-0.05, 0.05))), 2)

    data.append([
        score,
        time_taken,
        reattempts,
        topic_errors,
        avg_difficulty,
        consistency,
        level
    ])

df = pd.DataFrame(data, columns=[
    "score",
    "time_taken",
    "reattempts",
    "topic_errors",
    "avg_difficulty",
    "consistency",
    "level"
])


df.to_csv("custom_learning_data_1000.csv", index=False)

print("✅ 1000-row dataset created successfully!")
```
